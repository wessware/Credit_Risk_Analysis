```python
%%sql

SELECT COUNT(*) AS Total_Records from risk_analysis_dataset
```

```python
%%sql

SELECT Age, COUNT(*) as Number_of_Customers

FROM risk_analysis_dataset
GROUP BY Age
ORDER BY Age;
```

```python
%%sql

SELECT Annual_Income, Monthly_Inhand_Salary, (Monthly_Inhand_Salary * 12) AS Calculated_Annual_Income, (Annual_Income - (Monthly_Inhand_Salary * 12)) as Income_discrepancy

FROM risk_analysis_dataset;
```

```python
%%sql

SELECT Num_Bank_Accounts, COUNT(*) AS Frequency

FROM risk_analysis_dataset

GROUP BY Num_Bank_Accounts
ORDER BY Num_Bank_Accounts;
```

```python
%%sql

SELECT CASE
        WHEN Annual_Income < 30000 THEN "LOW"
        WHEN Annual_Income BETWEEN 30000 AND 70000 THEN "MID"
        ELSE "HIGH"
        END AS Income_tier,
        AVG(Num_Credit_Card) AS Averagee_Number_of_Cards
FROM risk_analysis_dataset
GROUP BY Income_tier;
```

```python
%%sql

SELECT MIN(Interest_Rate) AS Minimum_Rate,
        MAX(Interest_Rate) AS Maximum_Rate,
        AVG(Interest_Rate) AS Average_Rate

FROM risk_analysis_dataset;
```

```python
%%sql

SELECT Num_of_Loan, AVG(Num_of_Delayed_Payment) AS Average_delays

FROM risk_analysis_dataset

GROUP BY Num_of_Loan
ORDER BY Num_of_Loan;
```

```python
%%sql

SELECT Credit_Mix, AVG(Outstanding_Debt) AS Average_Debt

FROM risk_analysis_dataset

GROUP BY Credit_Mix;
```

```python
%%sql 

SELECT Payment_Behaviour, COUNT(*) AS Frequency

FROM risk_analysis_dataset

GROUP BY Payment_Behaviour
ORDER BY Frequency DESC;
```

```python
%%sql

SELECT CASE
        WHEN Annual_Income < 30000 THEN "LOW"
        WHEN Annual_Income BETWEEN 30000 AND 70000 THEN "MID"
        ELSE "HIGH"
        END AS Income_tiers,
        AVG(Credit_Utilization_Ratio) AS Average_Utilization

FROM risk_analysis_dataset

GROUP BY Income_tiers;
```

```python
%%sql

SELECT Payment_Behaviour, AVG(Credit_History_Age) AS Average_Credit_History

FROM risk_analysis_dataset

GROUP BY Payment_Behaviour;
```

```python
%%sql 

SELECT Amount_invested_monthly, AVG(Monthly_Balance) AS Average_monthly_balance

FROM risk_analysis_dataset

GROUP BY Amount_invested_monthly
ORDER BY Amount_invested_monthly;
```

```python
%%sql

SELECT Annual_Income, Outstanding_Debt, (Outstanding_Debt / Annual_Income) AS Debt_to_Income_Ratio

FROM risk_analysis_dataset;
```

```python
%%sql

SELECT Monthly_Inhand_Salary, Total_EMI_per_month, (Total_EMI_per_month / Monthly_Inhand_Salary) AS EMI_burden_Ratio 

FROM risk_analysis_dataset;
```

```python
%%sql

SELECT Num_Of_Delayed_Payment, Delay_from_due_date, (Num_of_Delayed_Payment * Delay_from_due_date) AS Severity_Index

FROM risk_analysis_dataset;
```

```python
%%sql

SELECT Num_Credit_Inquiries, AVG(Changed_Credit_Limit) AS Average_limit_change

FROM risk_analysis_dataset

GROUP BY Num_Credit_Inquiries
ORDER BY Num_Credit_Inquiries ASC;
```

```python
%%sql 

SELECT Payment_of_Min_Amount, COUNT(*) AS Number_of_Customers

FROM risk_analysis_dataset

GROUP BY Payment_of_Min_Amount;
```

```python
%%sql

SELECT *,
        CASE
            WHEN Credit_Utilization_Ratio > 0.8
                AND Num_of_Delayed_Payment > 5
                AND Outstanding_Debt > Annual_Income * 0.5
            THEN "HiGH RISK"
            ELSE "LOW/MEDIUM RISK"
        END AS Credit_risk_category
        
FROM risk_analysis_dataset;
```

```python
%%sql

SELECT Annual_Income, Amount_invested_monthly, (Amount_invested_monthly * 12 / Annual_Income) AS Income_to_Investment_Ratio

FROM risk_analysis_dataset

ORDER BY Income_to_Investment_Ratio DESC;
```

```python
%%sql

SELECT CASE
        WHEN (Outstanding_Debt / Annual_Income) < 0.2 AND Credit_Utilization_Ratio < 0.2 THEN "HEALTHY"
        WHEN (Outstanding_Debt / Annual_Income) BETWEEN 0.2 AND 0.3 THEN "MODERATE"
        ELSE "STRESSED"
        END AS Financial_Health,
        COUNT(*) AS Number_of_Customers

FROM risk_analysis_dataset
GROUP BY Financial_Health;
```

```python
%%sql

SELECT Age, Num_of_Delayed_Payment,
        AVG(Num_of_Delayed_Payment) OVER (PARTITION BY Age ROWS BETWEEN 2 PRECEDING AND CURRENT ROW) AS Rolling_Average_Delays

FROM risk_analysis_dataset;
```

```python
%%sql

SELECT Annual_Income, Outstanding_Debt, (Outstanding_Debt / Annual_Income) as Customer_DTI_ratio

FROM risk_analysis_dataset
ORDER BY Customer_DTI_ratio DESC
LIMIT 10;
```
