```python
%%sql 

ALTER TABLE risk_analysis_dataset

ADD COLUMN debt_to_income_ratio DECIMAL(10, 4);
```

```python
%%sql

UPDATE risk_analysis_dataset

SET debt_to_income_ratio = 
    (
        GREATEST(0, LEAST(Outstanding_Debt / (Annual_Income+1), 1)) 
    )
```

```python
%%sql 

SELECT Annual_Income, debt_to_income_ratio from risk_analysis_dataset LIMIT 5;
```

```python
%%sql 

ALTER TABLE risk_analysis_dataset

ADD COLUMN normalized_emi DECIMAL(10, 6);
```

```python
%%sql

UPDATE risk_analysis_dataset

SET normalized_emi = (
    GREATEST(0, LEAST(Total_EMI_per_month / (Monthly_Inhand_Salary+1), 1))
);
```

```python
%%sql 

SELECT Annual_Income, normalized_emi from risk_analysis_dataset LIMIT 5;
```

```python
%%sql 

ALTER TABLE risk_analysis_dataset

ADD COLUMN normalized_delinquency DECIMAL(10,6);
```

```python
%%sql 

UPDATE risk_analysis_dataset 

SET normalized_delinquency = (
    GREATEST(0, LEAST(Num_of_Delayed_Payment / (Num_of_Loan+1), 1))
);
```
